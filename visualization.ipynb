```python
import matplotlib.pyplot as plt
import torch
```

```python
chkp = torch.load("log/model_256000.pt", weights_only=False, map_location=torch.device("cpu"))
```

```python

wpe = chkp["model"]["transformer.wpe.weight"]
```

```python
plt.figure(figsize = (20,2))
plt.imshow(wpe, cmap="inferno")
plt.show()
```

```python
similarity = wpe @ wpe.T
plt.imshow(similarity, cmap="inferno")
plt.show()
```

```python
wte = chkp["model"]["transformer.wte.weight"]
```

```python
plt.figure(figsize = (20,2))
plt.imshow(wte.T, cmap="inferno")
plt.show()
```

```python
similarity = wte.T @ wte
plt.imshow(similarity, cmap="inferno")
plt.show()
```
